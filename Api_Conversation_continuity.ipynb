```python
from openai import OpenAI

from openai_API import second_response
```

```python
client=OpenAI(
    api_key="")
model="gpt-4.1-mini"
```

```python
response=client.chat.completions.create(
    model=model,
)
print("chat completion responce:", response.choices[0].message.content)
conversation = [
    {"role": "user", "content": "Tell me a story about a dog."},
    {"role": "assistant", "content": response.choices[0].message.content}]

follwup=client.chat.copletion.create(
    model=model,
    message=conversation+[{"role": "user", "content":"Tell me a story about a dog."}]
)
print("follow-up completion responce:", follwup.choices[0].message.content)
```

```python
response=client.responses.create(
    model=model,
    input=[{"role":"user","content":"Tell me a story about a dog."}]
)
print("responses api:",response.output_text)

second_response=client.responses.create(
    model=model,
    previous_response_id=response.id,
    input=[{"role":"user","content":"Tell me a story about a dog."}]
)

print("chaid responses:",second_response.output_text)
```
